
## Question Filtering

### Introduction
Question filtering is the process of retrieving specific records from a dataset based on given conditions.

In our AI Interview Simulator, candidates should be able to select their preferred job domain, topic, and difficulty level.

For example, if a candidate selects:
- Domain: Data Analyst
- Topic: SQL
- Difficulty: Easy

The system should retrieve only the interview questions matching all three conditions.

### Pandas Filtering
We will use Boolean indexing and the `.loc[]` accessor to filter records.

Boolean conditions return True or False depending on whether a record satisfies the specified condition.


```python
filtered_df = df.loc[df['domain'] == 'Data Analyst']
```

```python
filtered_df.shape
```


### Filtering Using Multiple Conditions

We can filter a Pandas DataFrame using multiple conditions simultaneously.

The `&` operator represents logical AND in Pandas filtering. It ensures that all specified conditions must be True for a row to be selected.

We use parentheses around individual conditions to maintain correct operator precedence.


```python

sql_easy_questions = df.loc[
    (df['domain'] == 'Data Analyst') &
    (df['topic'] == 'SQL') &
    (df['difficulty'] == 'Easy')
]
```

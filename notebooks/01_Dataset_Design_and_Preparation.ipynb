# Dataset Design and Preparation

## 1. Objective

The objective of Phase 2 was to prepare a structured interview-question dataset for the AI-Powered Interview Readiness Assessment and Career Improvement Platform.

The dataset will support role-specific interviews, candidate-answer evaluation, readiness scoring, weak-area identification, and personalized recommendations in later phases.

## 2. Dataset Creation

Three CSV datasets were prepared, each containing 1,000 records.

### A. Questions Dataset — `questions.csv`

Contains interview questions and their metadata.

Columns:
- `question_id`
- `domain`
- `topic`
- `difficulty`
- `question`
- `source`

The dataset covers multiple domains, including Data Analyst, Data Scientist, Software Engineer, Python Developer, Frontend Developer, Backend Developer, Full Stack Developer, Business Analyst, Mechanical Engineer, Civil Engineer, Electrical Engineer, and HR & Behavioral.

Difficulty distribution:
- Easy: 400
- Medium: 400
- Hard: 200

### B. Expected Answers Dataset — `expected_answers.csv`

Contains the expected answer for each interview question.

Columns:
- `question_id`
- `question`
- `expected_answer`
- `answer_level`

Answer-level categories after correcting one inconsistent label:
- Basic: 370
- Intermediate: 370
- Advanced: 185
- Behavioral: 75

A detailed review of answer quality was postponed to a later stage.

### C. Keywords Dataset — `keywords.csv`

Contains keywords that can help evaluate a candidate's response.

Columns:
- `question_id`
- `question`
- `keywords`

Each question has approximately 5–9 keywords, with an average of about 6.45 keywords per entry.

Original keyword capitalization was preserved.

## 3. Data Loading and Inspection

- Loaded the three CSV datasets into Pandas DataFrames.
- Inspected DataFrame dimensions, columns, and data types.
- Checked for missing values and blank strings.
- Checked for completely duplicated records.
- Checked for duplicate question IDs.
- Verified that question IDs ranged from 1 to 1,000 without gaps.
- Verified that the IDs and row order matched across all three datasets.
- Checked question-text consistency across the datasets.
- Corrected five question-text punctuation mismatches in the expected-answers DataFrame.
- Corrected one inconsistent answer-level label from `Medium` to `Intermediate`.

The expected-answers CSV initially required a different encoding to load successfully. The cleaned copy was subsequently saved using UTF-8 encoding.

## 4. Saving the Cleaned Datasets

Created three cleaned CSV files without overwriting the original files:

- `questions_cleaned.csv`
- `expected_answers_cleaned.csv`
- `keywords_cleaned.csv`

Saved the files in:

`app/data/`

Reloaded the saved files and confirmed their dimensions:

- Questions: 1,000 rows × 6 columns
- Expected answers: 1,000 rows × 4 columns
- Keywords: 1,000 rows × 3 columns

## 5. Dataset Integration

Merged the three DataFrames using `question_id`.

The combined DataFrame is named `interview_dataset`.

Its columns are:
1. `question_id`
2. `domain`
3. `topic`
4. `difficulty`
5. `question`
6. `source`
7. `expected_answer`
8. `answer_level`
9. `keywords`

### Final validation results

- Total records: 1,000
- Total columns: 9
- Duplicate question IDs: 0
- Missing values: 0
- Question IDs aligned across all three source datasets: Yes

## 6. Final Output

Saved the combined dataset as:

`app/data/interview_dataset.csv`

The three cleaned source datasets were retained separately.

## 7. Phase 2 Outcome

The project now has a structured, validated dataset that can be used by the future interview-question selection and answer-evaluation components.

Phase 2 is complete for the current development milestone. A full manual review of the correctness and depth of every expected answer has not yet been performed and may be done later if testing identifies problems.
